# Correction delivery timeline
Run from `lib/um982py` with matplotlib installed. The demonstration uses the checked-in sample capture, so replace its paths with your own verified data.


```python
import csv
import tempfile
from pathlib import Path

from examples.b2b_correction_audit.__main__ import main

work = Path(tempfile.mkdtemp(prefix="um982-science-"))
result = main(
    [
        "--replay",
        "tests/fixtures/specialist/labelled-b2b.bin",
        "--navigation",
        "tests/fixtures/specialist/navigation-issues.csv",
        "--json",
        str(work / "report.json"),
        "--checks-csv",
        str(work / "checks.csv"),
    ]
)
if result != 0:
    raise RuntimeError(f"Example returned status {result}")
print("Results:", work)
```

```python
import matplotlib.pyplot as plt

with (work / "checks.csv").open() as stream:
    rows = list(csv.DictReader(stream))
for kind in sorted({r["kind"] for r in rows}):
    data = [r for r in rows if r["kind"] == kind]
    plt.scatter(
        [float(r["received_gps_seconds"]) for r in data],
        [float(r["age_seconds"]) for r in data],
        label=kind,
    )
plt.ylabel("Correction age (s)")
plt.xlabel("GPS seconds since 1980-01-06")
plt.legend()
plt.show()
```
