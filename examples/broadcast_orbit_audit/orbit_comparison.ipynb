# Broadcast orbit versus SP3
Run from `lib/um982py` with matplotlib installed. The demonstration uses the checked-in sample capture, so replace its paths with your own verified data.


```python
import csv
import tempfile
from pathlib import Path

from examples.broadcast_orbit_audit.__main__ import main

work = Path(tempfile.mkdtemp(prefix="um982-science-"))
result = main(
    [
        "--replay",
        "tests/fixtures/specialist/rtklib-motion.bin",
        "--sp3",
        "tests/fixtures/specialist/rtklib-motion.sp3",
        "--json",
        str(work / "report.json"),
        "--differences-csv",
        str(work / "rac.csv"),
    ]
)
if result != 0:
    raise RuntimeError(f"Example returned status {result}")
print("Results:", work)
```

```python
import matplotlib.pyplot as plt

with (work / "rac.csv").open() as stream:
    rows = list(csv.DictReader(stream))
for sv in sorted({r["satellite"] for r in rows}):
    data = [r for r in rows if r["satellite"] == sv]
    plt.plot(
        [float(r["gps_seconds"]) for r in data],
        [float(r["radial_m"]) for r in data],
        label=sv,
    )
plt.ylabel("Radial difference (m)")
plt.xlabel("GPS seconds since 1980-01-06")
plt.legend()
plt.show()
```
